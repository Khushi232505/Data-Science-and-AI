#***Building a Neural Network***

Building a Neural Network is the process of creating the structure of ANN by adding layers, neurons, and activation functions using TensorFlow and Keras.

#***Steps to Build a Neural Network***

Import Libraries -> Create Sequential Model -> Add Input Layer -> Add Hidden Layer(s) -> Add Output Layer -> Compile Model -> Train Model -> Make Predictions

#***Components of a neural network***

1. Input Layer
2. Output Layer
3. Hidden Layer

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

#Create Neural Network
model = Sequential()

#Input + Hidden Layer
model.add(Dense(
    units=16,
    activation = "relu",
    input_shape=(2,)
))

#Hidden Layer
model.add(Dense(
    units=8,
    activation = "relu"
))

#Ouput Layer
model.add(Dense(
    units=1,
    activation = "sigmoid"
))

#Display Model Summary
model.summary()
```

#***Model Compilation in TensorFlow and Keras***
Model Compilation is the process of configuring a nural network by specifying the optimizer, loss function, and evaluation metrics before training.

#***Components:***

1. ***Optimizer*** - An optimizer adjusts the weights of the neural network to minimize the loss and improve prediction accuracy
* adam
* SGD
* RMSprop
* Adagrad

2. ***Loss Function*** - The loss function calculates the prediction error of the neural network.
* binary_crossentropy
* categorical_crossentrpy
* mean_squared_error(MSE)

3. ***Metrics*** - Metrics evaluate how well the neural network is performing.

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

#Create Model
model = Sequential()

#Input + Hidden Layer
model.add(Dense(16, activation = "relu", input_shape = (2,)))

#Hidden Layer
model.add(Dense(8, activation = "relu"))

#Output Layer
model.add(Dense(1, activation = "sigmoid"))

#Compile Model
model.compile(
    optimizer = "adam",
    loss = "binary_crossentropy",
    metrics = ["accuracy"]
)

model.summary()
```

#***Artificial Neuron***

An Artificial Neuron is the mathematical model of a biological neuron used in Artificial Neural Networks(ANNs). It receives input values, performs
calculations, applies an activation function, and produces an output.

#***Mapping Between Biological and Artificial Neurons.***

#***Biological Neuron Artificial Neuron***

1. Dendrites ---------> Inputs(X)
2. Synapse -----------> Weights(W)
3. Cell Body ---------> Weighted Sum(Σ)
4. Axon -----------> Output(Y)
5. Brain ---------> Neural Network(ANN)

#***Actiavtion Functions***

An activation function determines the output of a neuron by transforming the weighted sum into a meaningful value.

Inputs(X) --> Weighted Sum(Σ) --> Activation Function --> Output

#***Types of Activation Functions***

1. Binary Step -- The binary step function gives only two outputs(0,1).It is mainly used in perception model

2. Sigmoid Function -- It converts any input into a value between 0 and 1.

3. Tanh Function -- The tanh functions output values between -1 and +1.

4. ReLU(Rectified Linear Unit) -- It is used activation function in deep learning

5. Leaky ReLU -- Leaky ReLU is an improved version of ReLU. Instead of returning 0 for negative values, it returns a small negative value.

6. Softmax -- Softmaxx is used for Multi-Class Classification .
It converts outputs into probabilties whose total is 1.

#***Perception Model***

A Perception is the simplest Artificial Neural Network(ANN) and is the building block of deep learning. It was introduced by Frank Rosenblatt in 1958.

It is mainly used for Binary Classification, where the output has only two classes:

1. Yes/No
2. Pass/Fail
3. Spam/Not Spam
4. Survived/Not Survived

```python
import numpy as np

# Inputs
hours = 8
attendance = 90

# Weights
w1 = 0.6
w2 = 0.4

# Bias
bias = -4

# Weighted Sum
z = (hours  * w1) + (attendance * w2) + bias

print("Weighted Sum : ", z)

# Step Activation Function
if z>= 0:
  print("Prediction : Pass")
else:
  print("Prediction : Fail")
```

#***Feed Forward Neural Network(FFNN)***

A Feed Forward Neural Network is a neural network where information moves only from the input layer to the output layer without any backward connections.

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

#Create Feed Forward Neural Netwrok
model = Sequential()

#Input + Hidden Layer
model.add(Dense(16, activation = "relu", input_shape = (7,)))

#Hidden Layer
model.add(Dense(8, activation = "relu"))

#Output Layer
model.add(Dense(1, activation = "sigmoid"))

# Display Model Architecture
model.summary()

```
