```python
#Practical Example 1: Creating Tensors
import tensorflow as tf

#Scalar Tensor (0D)
scalar = tf.constant(10)
print("Scalar:", scalar)

#Vector Tensor (1D)
vector = tf.constant([10, 20, 30])
print("Vector:", vector)

# Matrix Tensor (2D)
matrix = tf.constant([
    [1,2],
    [3,4]
])
print("Matrix : ")
print(matrix)
```

```python
#Practical Example 2:Tensors Operations
import tensorflow as tf

a = tf.constant([[1,2],[3,4]])
b = tf.constant([[5,6],[7,8]])

print("Addition:", tf.add(a, b))
print("Subtraction:", tf.subtract(a, b))
print("Multiplication:", tf.multiply(a, b))
print("Division:", tf.divide(a, b))
```

```python
#Practical Example 3: Matrix Multiplication

import tensorflow as tf

A = tf.constant([
    [1,2],
    [3,4]
])

B = tf.constant([
    [5,6],
    [7,8]
])

result = tf.matmul(A,B)
print(result)
```

#***Building a Neural Network***

Building a Neural Network is the process of creating the structure of ANN by adding layers, neurons, and activation functions using TensorFlow and Keras.

#***Steps to Build a Neural Network***

Import Libraries -> Create Sequential Model -> Add Input Layer -> Add Hidden Layer(s) -> Add Output Layer -> Compile Model -> Train Model -> Make Predictions

#***Components of a neural network***

1. Input Layer
2. Output Layer
3. Hidden Layer

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

#Create Neural Network
model = Sequential()

#Input + Hidden Layer
model.add(Dense(
    units=16,
    activation = "relu",
    input_shape=(2,)
))

#Hidden Layer
model.add(Dense(
    units=8,
    activation = "relu"
))

#Ouput Layer
model.add(Dense(
    units=1,
    activation = "sigmoid"
))

#Display Model Summary
model.summary()
```

#***Model Compilation in TensorFlow and Keras***
Model Compilation is the process of configuring a nural network by specifying the optimizer, loss function, and evaluation metrics before training.

#***Components:***

1. ***Optimizer*** - An optimizer adjusts the weights of the neural network to minimize the loss and improve prediction accuracy
* adam
* SGD
* RMSprop
* Adagrad

2. ***Loss Function*** - The loss function calculates the prediction error of the neural network.
* binary_crossentropy
* categorical_crossentrpy
* mean_squared_error(MSE)

3. ***Metrics*** - Metrics evaluate how well the neural network is performing.

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

#Create Model
model = Sequential()

#Input + Hidden Layer
model.add(Dense(16, activation = "relu", input_shape = (2,)))

#Hidden Layer
model.add(Dense(8, activation = "relu"))

#Output Layer
model.add(Dense(1, activation = "sigmoid"))

#Compile Model
model.compile(
    optimizer = "adam",
    loss = "binary_crossentropy",
    metrics = ["accuracy"]
)

model.summary()
```
