# ***Matplotlib***

It is the most popular python library used for creating data visualizations.
It converts numerical data into meaningful charts and graphs that helps users analyze trends, patterns, and comparisons.

# ***Features of Matplotlib***

1. Open source
2. Highly customizable
3. Supports 20+ chart types
4. Publication Quality charts
5. Works perfectly with Numpy and Pandas
6. Can save images in PNG, JPG, PDF, SVG
