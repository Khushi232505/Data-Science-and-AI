# ***Pandas***

Pandas is an open source python library used for data manipulation and data analysis.

# ***Why do we use Pandas?***

Pandas is used to read, clean, organize, filter, analyze, and manipulate structured data efficiently.

# ***Features of Pandas : ***

# ***1. Data Manipulation***

Used to modify, organize, and transform data easily.

# ***2. Data Analysis***

Used to analyze data and extract meaningful insights.

# ***3. Fast Processing***

Handles large datasets efficiently with optimized performance.

# ***4. Missing Value Handling***

Allows easy detection and handling os missing values.

# ***5. File Handling***

Can read and write CSV, Excel, JSON, Sql and many other files.

# ***6. Data Filtering***

Used to select specific rows and columns based on conditions

Installation of Pandas


```python
import pandas as pd
```

# ***Main Data Structures in Pandas***

1. series --> A series is a one-dimensional labeled array that stores a single column of data.

2. Data Frame --> A data frame is a two dimensional labeled table consisting of rows and columns

# ***1. Creating Series***

A series is a dimensional labeled array that stores a single column of data.

```python
import pandas as pd

marks = pd.Series([85, 90, 78, 92, 88])

print(marks)
```

Creating a series of strings

```python
import pandas as pd
names = pd.Series(["khushi", "rosu", "sejal", "pum pum"])
print(names)
```

# ***Custom Index***

```python
import pandas as pd

marks = pd.Series(
    [85, 90, 78],
    index = ["Maths", "hindi", "english"]
)

print (marks)
```

# ***Series From Dictionary***

```python
import pandas as pd

student = pd.Series({
    "name" : "khushi",
    "age" : 21,
    "city" : "lucknow"
})

print(student)
```

# ***Access Values in a Series***

```python
import pandas as pd

marks = pd.Series([85,90,78, 92, 88])

print(marks[0])
print(marks[2])
```

# ***Access Value Using Custom Index.***

```python
import pandas as pd

marks = pd.Series(
    [85, 90, 78],
    index = ["Math", "Hindi", "English"]
)

print(marks["Hindi"])
```

# ***Print Series Information***

```python
import pandas as pd

marks = pd.Series([85, 90, 78])

print(type(marks))
print(marks.dtype)
print(marks.size)
```

```python
import pandas as pd
student = { "Name": ["Aman", "Ravi", "Priya", "Neha"],
           "Age": [20, 21, 19, 22],
            "City": ["Delhi", "Lucknow", "Noida", "Jaipur"] }
df = pd.DataFrame(student)
print(df)
```

Create DataFrame with Custom Index

```python
import pandas as pd
student = { "Name": ["Aman","Ravi","Priya"],
           "Marks": [85,90,78] }
df = pd.DataFrame(student, index=["S1", "S2", "S3"])
print(df)
```
